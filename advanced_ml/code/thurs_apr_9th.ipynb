```python
X = np.array(data)
X, X.shape
```

```python
x_bar = np.mean(X,axis=0)
x_bar
```

```python
Y = X - x_bar
Y
```

```python
X_cov = np.cov(X, rowvar=False)
S_diag = np.diag(X_cov)

Z = Y / np.sqrt(S_diag)
Z
```

```python
D = np.eye(24) / 23
R = Z.T @ D @ Z
R_df = pd.DataFrame(R, columns=["Cylinder","Power","Velocity","Weight","Length","Width"])
R_df
```

```python
eigenvalues, eigenvectors = np.linalg.eig(R)
eigenvectors
```

```python
C = Z @ eigenvectors
C
```
