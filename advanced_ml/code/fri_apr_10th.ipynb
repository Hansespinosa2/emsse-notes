```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
X = np.array([[2, 5],
     [4, 2],
     [6, 3],
     [3, 4],
     [5, 1]])
n, p = X.shape

x_bars = X.mean(axis=0)

x_vars = X.var(axis=0)

S = np.cov(X, rowvar=False, bias=True)

X, n, p, x_bars, x_vars
```

```python
var_vec = np.std(X,axis=0)

Z = (X - x_bars ) / var_vec

D_inv = np.diag(1/var_vec)

R = D_inv @ S @ D_inv
Z, R
```

```python
v, U = np.linalg.eig(R)

C = Z @ U
C
```
