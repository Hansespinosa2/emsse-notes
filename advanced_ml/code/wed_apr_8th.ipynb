```python
import numpy as np

y = np.array([47, 76, 97, 107, 123, 136, 152, 159, 191, 201, 200, 207])
x = np.array([0.02, 0.02, 0.06, 0.06, 0.11, 0.11, 0.22, 0.22, 0.56, 0.56, 1.1, 1.1])
x.shape, y.shape
```

```python
x = 1/x
y = 1/y
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
x_bar = np.mean(x)
y_bar = np.mean(y)

B_1 = np.dot(x-x_bar, y) / np.dot(x-x_bar, x-x_bar)

B_0 = y_bar - B_1 * x_bar

B_0, B_1
```

```python
theta_1 = 1 / B_0
theta_2 = B_1 / B_0

theta_1, theta_2
```

```python
y_pred = B_0 + B_1 * x
y_pred
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.plot(x, y_pred, color='red')
```

```python
import pandas as pd
df = pd.DataFrame({"x": x, "y": y, "y_pred": y_pred})
df
```

```python
SSR = np.sum((y_pred - y_bar)**2)
SSE = np.sum((y- y_pred)**2)
SST = np.sum((y-y_bar)**2)

SSR, SSE, SST
```

```python
n = np.shape(x)[0]
F = (n-2) * SSR / SSE 
F
```

```python
S_xx = np.sum((x-x_bar)**2)
S_sqr = np.sum((y-y_pred)**2 / (n-2))
S_xx, S_sqr
```
