### Exercise 7

```python
X = np.array([
    [167, 1, 163, 23, 41, 8, 6, 6],
    [162, 2, 141, 12, 40, 12, 4, 15],
    [119, 6, 69, 56, 39, 5, 13, 41],
    [87, 11, 63, 111, 27, 3, 18, 39],
    [103, 5, 68, 77, 32, 4, 11, 30],
    [111, 4, 72, 66, 34, 6, 10, 28],
    [130, 3, 76, 52, 43, 7, 7, 16],
    [138, 7, 117, 74, 53, 8, 12, 20]
])

col_names = ["bread_ord", "bread_oth", "wine_ord", "wine_oth", "potatoes", "veg", "grapes", "dishes"]

df = pd.DataFrame(X, columns=col_names)
df
```

```python
# First Part 
x_bar_vec = X.mean(axis=0)
Y = X - x_bar_vec

std_vec = X.std(axis=0)
D_inv = np.diag(1/std_vec)

Z = Y @ D_inv
Z
```

```python
# Second Part
S = np.cov(X, rowvar=False, bias=True)

R = D_inv @ S @ D_inv
R
```

```python
# Third Part
v, U = np.linalg.eig(R)
v
```

```python
col_with_eig = pd.Series({col_name: eig for col_name, eig in zip(col_names, v)})
col_with_eig.sort_values(ascending=False, inplace=True)

df_summary = pd.DataFrame(col_with_eig, columns=["eigenvalue"])
df_summary["cumsum"] = df_summary.cumsum()
total_sum = df_summary["cumsum"].max()
df_summary["cumulative_percentage"] = df_summary["cumsum"] / total_sum
df_summary
```

```python
C = Z @ U[:, :2]
C
```
