```python
# P-invariants and T-Invariants
import sympy as sp

def invariants(C):
    C = sp.Matrix(C)

    T = C.nullspace()

    P = C.T.nullspace()

    return P, T


# Example
C = [
    [-1, -1, 1, 1],
    [1, 0, -1, 0],
    [0, 1, 0, -1],
    [0, 0, 1, 0]
]

P, T = invariants(C)

print("P-invariants:", P)
print("T-invariants:", T)
```

```python
# Verifying
import numpy as np 
C_arr = np.array(C)
P_arr = np.array(P)
T_arr = np.array(T)
C_arr.T @ P_arr, C @ T_arr
```
